```python
from surprise import Dataset, SVD, SVDpp, NMF
from surprise.model_selection import GridSearchCV
```

```python
data = Dataset.load_builtin("ml-100k")
```

```python
param_grid = {"n_epochs": [5, 10], "lr_all": [0.002, 0.005], "reg_all": [0.4, 0.6]}
svd_gs = GridSearchCV(SVD, param_grid, measures=["rmse", "mae"], cv=3, n_jobs=-1)
svd_gs.fit(data)
```

```python
print('RMSE score:', svd_gs.best_score["rmse"])
print(svd_gs.best_params["rmse"])
```

```python
param_grid = {"n_factors": [5, 10, 20], "lr_all": [0.005, 0.007], "reg_all": [0.4, 0.6], "cache_ratings": [True]}
svdpp_gs = GridSearchCV(SVDpp, param_grid, measures=["rmse", "mae"], cv=3, n_jobs=-1)
svdpp_gs.fit(data)
```

```python
print('RMSE score:', svdpp_gs.best_score["rmse"])
print(svdpp_gs.best_params["rmse"])
```

```python
param_grid = {"n_factors": [5, 10, 20], "n_epochs": [10, 50, 100, 500], 'biased': [True, False]}
nmf_gs = GridSearchCV(NMF, param_grid, measures=["rmse", "mae"], cv=3, n_jobs=-1)
nmf_gs.fit(data)
```

```python
print('RMSE score:', nmf_gs.best_score["rmse"])
print(nmf_gs.best_params["rmse"])
```
