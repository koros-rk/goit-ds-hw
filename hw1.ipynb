Створіть наступний шаблон без жорсткого кодування (Hard coding). Використовуйте лише функції NumPy і наведений нижче масив введення.

Input:

`a = np.array([1, 2, 3])`

Output:

`array([1, 1, 1, 2, 2, 2, 3, 3, 3, 1, 2, 3, 1, 2, 3, 1, 2, 3])`

```python
a = np.array([1, 2, 3])

part_1 = np.append([], [
    np.repeat(n, a.size)
    for n in a
])
part_2 = np.tile(a, a.size)


np.append(part_1, part_2)
```


Знайдіть повторювані записи від 2-го випадку повторення і далі у заданому масиві і позначте їх як `True`. Перше повторення має бути `False`.

Input:

`np.random.seed(100)`

`a = np.random.randint(0, 5, 10)`

`[0 0 3 0 2 4 2 2 2 2]`

Output:

`[False True False True False False True True True True]`


```python
np.random.seed(100)
a = np.random.randint(0, 5, 10)
occurrences = []
result = []

for number in a:
    if number in occurrences:
        result.append(True)
    else:
        occurrences.append(number)
        result.append(False)

result
```
